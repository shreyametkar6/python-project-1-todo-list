```python
def add_task(tasks, text):
    text = text.strip()
    if not text:
        return False
    new_id = max((t["id"] for t in tasks), default = 0) + 1
    tasks.append({"id": new_id, "task": text})
    return True

def view_tasks(tasks):
    if not tasks:
        print("No tasks yet.")
        return
    for number, t in enumerate(tasks, start = 1):
        print(f"{number}. {t['task']}")
```

```python
def show_menu():
    print("\n1. Add Task")
    print("2. View Tasks")
    print("3. Quit")
    
def main():
    my_tasks = load_tasks()
    
    while True:
        show_menu()
        choice = input("Choose (1 - 3): ").strip()
        
        if choice == "1":
            if add_task(my_tasks, input("Enter Task: ")):
                save_tasks(my_tasks)
                print("Task Added!")
            else:
                print("Task can't be empty.")
        
        elif choice == "2":
            view_tasks(my_tasks)

        elif choice == "3":
            print("Good Bye!")
            break
        
        else:
            print("Invalid Choice.")
```

```python
if __name__ == "__main__":
    main()
```
